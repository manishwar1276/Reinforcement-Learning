```python
import random

machines = ['A', 'B', 'C']
epsilon = 0.2
plays = 3

N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': [], 'B': [], 'C': []}
Q = {'A': 0, 'B': 0, 'C': 0}

for i in range(plays):
    r = random.random()

    if r < epsilon:
        machine = random.choice(machines)
        print("Exploration")
    else:
        machine = max(Q, key=Q.get)
        print("Exploitation")

    print("Machine:", machine)
    reward = float(input("Enter reward: "))

    N[machine] += 1
    R[machine].append(reward)

    Q[machine] = Q[machine] + (reward - Q[machine]) / N[machine]

print("\nN Table:", N)
print("R Table:", R)
print("Q Table:", Q)

print("Best Machine:", max(Q, key=Q.get))
```
