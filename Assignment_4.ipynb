```python
import math
machines=['A','B','C']
N_play=10
N={'A':0,'B':0,'C':0}
R={'A':0,'B':0,'C':0}
UCB={'A':0,'B':0,'C':0}
Q={'A':0,'B':0,'C':0}
rewards={'A':[3,4,2,5],'B':[6,7,5,8],'C':[4,3,5,4]}
index={'A':0,'B':0,'C':0}
for play in range(1,N_play+1):
    if play<=3:
        machine=machines[play-1]
    else:
        for m in machines:
            # Handle cases where N[m] might be 0 for new machines before they are played
            # Or if a machine has run out of its predefined rewards and thus N[m] hasn't been updated recently
            # This check is crucial before dividing by N[m]
            if N[m] > 0:
                UCB[m]=Q[m]+math.sqrt((2*math.log(play))/N[m])
            else:
                UCB[m] = float('inf') # Assign a high value to unplayed machines to encourage exploration
        machine=max(UCB,key=UCB.get)
    # Fix: Use modulo operator to cycle through the rewards list
    reward=rewards[machine][index[machine] % len(rewards[machine])]
    index[machine]+=1
    print("\nPlay",play)
    print("Play with machine",machine)
    print("Reward:",reward)
    N[machine]+=1
    R[machine]+=reward
    Q[machine]=Q[machine]+(reward-Q[machine])/N[machine]
    for m in machines:
        if N[m]>0:
            UCB[m]=Q[m]+math.sqrt((2*math.log(play))/N[m])
        else:
            UCB[m]=float('inf')
print("\nFINAL RESULTS")
print("Machine\tN\tR\tUCB\t\tQ")
for m in machines:
    print(f"{m}\t{N[m]}\t{R[m]}\t{UCB[m]:.3f}\t\t{Q[m]:.3f}")
print("Total reward received:",sum(rewards['A'])+sum(rewards['B'])+sum(rewards['C']))
```
