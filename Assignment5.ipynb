```python
import gymnasium as gym

def random_agent(env_name):
    env=gym.make(env_name)
    state,info=env.reset(seed=42)
    env.action_space.seed(42)
    rewards=[1,1,1,1]
    total_reward=0
    T=0
    terminated=False
    truncated=False

    print("\nEnvironment:",env_name)

    while T<4:
        action=env.action_space.sample()
        next_state,reward,terminated,truncated,info=env.step(action)
        total_reward+=rewards[T]
        T+=1
        state=next_state
        if terminated or truncated:
            break

    total_reward=sum(rewards[:T])
    average_reward=total_reward/T

    print("Total Reward:",total_reward)
    print("Average Reward:",average_reward)
    env.close()

random_agent("CartPole-v1")
random_agent("MountainCar-v0")
random_agent("FrozenLake-v1")
```
