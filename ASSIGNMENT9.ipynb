```python
import numpy as np
import pandas as pd

# System of linear equations for Assignment 1:
# (1 - 0.09)*V(S0) - 0.81*V(S1) = -1.0
# -0.09*V(S0) + 1*V(S1) = 89.9

A = np.array([
    [0.91, -0.81],
    [-0.09, 1.0]
])

b = np.array([-1.0, 89.9])

# Solve for State Values
v0, v1 = np.linalg.solve(A, b)

# Calculate Action Values
gamma = 0.9
q_s0_down = -1 + gamma * v1
q_s0_right = -1 + gamma * v0
q_s1_right = 100 + gamma * 0  # Goal state G has value 0
q_s1_up = -1 + gamma * v0

# Print Table for Assignment 1
assignment1_results = {
    "State/Action": ["V_pi(S0)", "V_pi(S1)", "Q_pi(S0, Down)", "Q_pi(S0, Right)", "Q_pi(S1, Right)", "Q_pi(S1, Up)"],
    "Calculated Value": [round(x, 2) for x in [v0, v1, q_s0_down, q_s0_right, q_s1_right, q_s1_up]]
}

df_a1 = pd.DataFrame(assignment1_results)
print("=== ASSIGNMENT 1 RESULTS ===")
display(df_a1)

# Verification check
print("\nVerification Checks:")
v0_check = 0.9 * q_s0_down + 0.1 * q_s0_right
v1_check = 0.9 * q_s1_right + 0.1 * q_s1_up
print(f"V_pi(S0) Left Hand Side: {v0:.4f} | Right Hand Side: {v0_check:.4f}")
print(f"V_pi(S1) Left Hand Side: {v1:.4f} | Right Hand Side: {v1_check:.4f}")
```

```python
# Policy Evaluation
import math

v_s0, v_s1 = 0.0, 0.0
gamma = 0.9
theta = 0.000001
iterations = 0

while True:
    iterations += 1
    # Update equations
    v_s0_new = 0.1 * (-1) + 0.9 * (-1 + gamma * v_s1)
    v_s1_new = 0.9 * 100 + 0.1 * (-1 + gamma * v_s0)

    delta = max(abs(v_s0_new - v_s0), abs(v_s1_new - v_s1))
    v_s0, v_s1 = v_s0_new, v_s1_new

    if delta < theta:
        break

# Calculate Action Values After Convergence
q2_s0_down = 0.1 * (-1 + gamma * 0) + 0.9 * (-1 + gamma * v_s1)
q2_s0_right = 1.0 * (-1 + gamma * 0)  # Hits wall
q2_s1_right = 0.9 * (100 + gamma * 0) + 0.1 * (-1 + gamma * v_s0)
q2_s1_up = 1.0 * (-1 + gamma * v_s0) # Moves to S0

# Display Results Table
assignment2_results = {
    "Metric / State / Action": [
        "Number of evaluation iterations",
        "V_pi(S0)",
        "V_pi(S1)",
        "Q_pi(S0, Down)",
        "Q_pi(S0, Right)",
        "Q_pi(S1, Right)",
        "Q_pi(S1, Up)"
    ],
    "Calculated Value": [
        iterations,
        round(v_s0, 6),
        round(v_s1, 6),
        round(q2_s0_down, 6),
        round(q2_s0_right, 6),
        round(q2_s1_right, 6),
        round(q2_s1_up, 6)
    ]
}

df_a2 = pd.DataFrame(assignment2_results)
print("=== ASSIGNMENT 2 RESULTS ===")
display(df_a2)

# Policy Consistency Verification
print("\nConsistency Checks:")
print(f"Is V_pi(S0) == Q_pi(S0, Down)? {math.isclose(v_s0, q2_s0_down, rel_tol=1e-5)}")
print(f"Is V_pi(S1) == Q_pi(S1, Right)? {math.isclose(v_s1, q2_s1_right, rel_tol=1e-5)}")
```
